## Boundary-impact metrics

`dropdown_edge` is the mean signed raw ligand-count difference across the 20%, 40%, 60%, and 80% edge-dropout results. It is not normalized by each tool's output scale. Larger positive values mean more calls were lost; negative values mean the perturbed result contained more calls.

`equal_edge` and `gradient_edge` are two-sample Kolmogorov-Smirnov (KS) statistics. The KS statistic is `D = max_x |F_original(x) - F_perturbed(x)|`, the largest vertical distance between the two empirical cumulative distribution functions. `D` ranges from 0 to 1: 0 means the observed distributions coincide, while a larger value means a stronger distribution shift. It does not identify the direction of change.

Because `dropdown_edge` is an unnormalized raw-count difference and the other two columns are KS statistics, the three columns must not be averaged or interpreted as one common-scale boundary-adoption score.
