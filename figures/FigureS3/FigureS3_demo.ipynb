## Result contract

Reads the latest existing DLPFC/MF h5ad objects; no Moran/Geary recalculation. A source manifest verifies that the cached script panels postdate those inputs. The final redraw is emitted as two separate 3-column by 4-row SVGs with receptor/ligand metrics on the left and interaction panels on the right; panel letters are omitted.

`results/generated/` contains the latest script-generated SVG/CSV/JSON files. `results/published/` contains the submitted assembly and may include manual layout.


## Reproduce

This cell calls the same figure shortcut used from the command line. Figure2 runs its repository-contained CDF step; Figure3-5 use their canonical latest inputs. FigureS3 reads precomputed h5ad files from `SPATIALCCCBENCH_DATA_ROOT`.


```python
from pathlib import Path
import os
import subprocess
import sys

figure = 'FigureS3'
script = Path.cwd() / 'run.py'
if not script.exists():
    script = Path('figures') / figure / 'run.py'
command = [sys.executable, str(script), '--reproduce']
data_root = Path(os.environ['SPATIALCCCBENCH_DATA_ROOT'])
command.extend(['--data-root', str(data_root)])
subprocess.run(command, check=True)
```
